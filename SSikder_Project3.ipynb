```python
query = """
SELECT 
    c.LastName,
    c.FirstName,
    t.Name AS TrackName,
    a.Title AS AlbumTitle
FROM Customer c
JOIN Invoice i
    ON c.CustomerId = i.CustomerId
JOIN InvoiceLine il
    ON i.InvoiceId = il.InvoiceId
JOIN Track t
    ON il.TrackId = t.TrackId
JOIN Album a
    ON t.AlbumId = a.AlbumId
ORDER BY c.LastName, c.FirstName;
"""

df = pd.read_sql(query, conn)

df.head()
```
