## Data Loading
In this step, we load the Auto MPG dataset from an online source into a pandas DataFrame and assign appropriate column names.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "http://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data"

columns = ["mpg", "cylinders", "displacement", "horsepower",
           "weight", "acceleration", "model_year", "origin", "car_name"]

df = pd.read_csv(url, names=columns, delim_whitespace=True, na_values='?')
```

```python
df.head()
```

## Data Cleaning
In this step, we handle missing values in the horsepower column and convert it to numeric. We also convert the origin column into meaningful labels.

```python
df['horsepower'] = pd.to_numeric(df['horsepower'])
```

```python
df['origin'] = df['origin'].map({1: 'USA', 2: 'Asia', 3: 'Europe'})
```

```python
df.head()
```

## Distribution of Cylinders
This bar chart shows how many cars fall into each cylinder category.

```python
plt.figure()
df['cylinders'].value_counts().sort_index().plot(kind='bar')
plt.title("Distribution of Cylinders")
plt.xlabel("Cylinders")
plt.ylabel("Count")
plt.show()
```

## Horsepower vs Weight
This scatterplot shows the relationship between horsepower and vehicle weight.

```python
plt.figure()
sns.scatterplot(data=df, x='horsepower', y='weight')
plt.title("Horsepower vs Weight")
plt.xlabel("Horsepower")
plt.ylabel("Weight")
plt.show()
```

## Question:
Do cars from different regions have different fuel efficiency (MPG)?

```python
plt.figure()
sns.boxplot(data=df, x='origin', y='mpg')
plt.title("MPG Distribution by Origin")
plt.xlabel("Origin")
plt.ylabel("MPG")
plt.show()
```

## Answer:
The graph shows that fuel efficiency (MPG) varies by region. Cars from Asia generally have higher MPG compared to cars from the USA and Europe. This suggests that Asian cars are more fuel-efficient. Cars from the USA tend to have lower MPG, likely because they are heavier and have larger engines.
