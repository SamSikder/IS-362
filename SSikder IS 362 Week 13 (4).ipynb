```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Loading the Dataset

The mushroom dataset was imported directly from the UCI Machine Learning Repository. This dataset contains observations about different mushroom characteristics such as odor, cap color, gill size, habitat, and whether the mushroom is edible or poisonous.

Loading the data directly from the source ensures the notebook can be reproduced without needing local files.

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/mushroom/agaricus-lepiota.data"

columns = [
    "class", "cap_shape", "cap_surface", "cap_color", "bruises", "odor",
    "gill_attachment", "gill_spacing", "gill_size", "gill_color",
    "stalk_shape", "stalk_root", "stalk_surface_above_ring",
    "stalk_surface_below_ring", "stalk_color_above_ring",
    "stalk_color_below_ring", "veil_type", "veil_color",
    "ring_number", "ring_type", "spore_print_color",
    "population", "habitat"
]

mushrooms = pd.read_csv(url, header=None, names=columns)
mushrooms.head()
```

## Selecting Relevant Variables

For this analysis, I selected three variables:

- **class**: Indicates whether the mushroom is edible or poisonous  
- **odor**: Describes the smell of the mushroom  
- **cap_color**: Describes the mushroom cap color

These variables were chosen because odor is commonly associated with spoilage or toxicity, while cap color may also contain useful visual information.

```python
# Select needed columns
df = mushrooms[["class", "odor", "cap_color"]].copy()

# Show first rows
df.head()
```

## Converting Categorical Values to Numeric Codes

The original dataset uses single-letter codes to represent categories. Since scikit-learn algorithms generally require numeric values, these categories were converted into integers using dictionary mapping.

For example:

- edible = 0  
- poisonous = 1

This transformation keeps the dataset machine-learning ready while preserving category distinctions.

```python
# Convert letters to numbers

df["class"] = df["class"].map({"e": 0, "p": 1})

df["odor"] = df["odor"].map({
    "a": 0, "l": 1, "c": 2, "y": 3,
    "f": 4, "m": 5, "n": 6, "p": 7, "s": 8
})

df["cap_color"] = df["cap_color"].map({
    "n": 0, "b": 1, "c": 2, "g": 3,
    "r": 4, "p": 5, "u": 6, "e": 7,
    "w": 8, "y": 9
})

df.head()
```

## Exploratory Data Analysis

The next step is to visualize the selected variables. Frequency distributions help identify common categories, while scatterplots allow comparison between mushroom class and predictor variables.

These visuals can reveal patterns that may be useful in future classification models.

```python
# Distribution of class
df["class"].value_counts().plot(kind="bar")
plt.title("Edible vs Poisonous")
plt.xlabel("0 = Edible, 1 = Poisonous")
plt.ylabel("Count")
plt.show()

# Distribution of odor
df["odor"].value_counts().sort_index().plot(kind="bar")
plt.title("Odor Distribution")
plt.xlabel("Odor Code")
plt.ylabel("Count")
plt.show()

# Distribution of cap color
df["cap_color"].value_counts().sort_index().plot(kind="bar")
plt.title("Cap Color Distribution")
plt.xlabel("Cap Color Code")
plt.ylabel("Count")
plt.show()
```

```python
# Poisonous/Edible vs Odor
plt.scatter(df["odor"], df["class"])
plt.title("Class vs Odor")
plt.xlabel("Odor")
plt.ylabel("Class")
plt.show()

# Poisonous/Edible vs Cap Color
plt.scatter(df["cap_color"], df["class"])
plt.title("Class vs Cap Color")
plt.xlabel("Cap Color")
plt.ylabel("Class")
plt.show()
```

## Conclusions

Based on the visual analysis, **odor appears to be the strongest predictor** of whether a mushroom is edible or poisonous. The scatterplot shows clearer separation between classes across odor categories, suggesting that certain smells are strongly associated with poisonous mushrooms.

Cap color shows weaker predictive power. While some colors may appear more often in one class, there is substantial overlap between edible and poisonous mushrooms, making color alone less reliable.

If a classification model such as Decision Tree, Logistic Regression, or Random Forest were built in scikit-learn, odor would likely be one of the most important features included in the model.

Overall, this assignment demonstrates the importance of preprocessing categorical data into numeric form before machine learning can be applied.
